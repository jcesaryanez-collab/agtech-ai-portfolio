```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Simulamos 200 registros de sensores de un tractor
np.random.seed(42)
n = 200

datos = pd.DataFrame({
    "horas_uso": np.arange(1, n + 1),
    "temp_motor_c": np.random.normal(85, 5, n).round(1),
    "vibracion_mm_s": np.random.normal(4.5, 1.2, n).round(2),
    "falla": np.random.choice([0, 1], n, p=[0.85, 0.15])
})

datos.head()
```

```python
# Resumen estadístico de las variables
datos.describe().round(2)
```

```python
# ¿La vibración es mayor cuando hay falla?
datos.groupby("falla")["vibracion_mm_s"].mean().round(2)
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(datos["temp_motor_c"], datos["vibracion_mm_s"],
            c=datos["falla"], cmap="coolwarm", alpha=0.7)
plt.xlabel("Temperatura del motor (°C)")
plt.ylabel("Vibración (mm/s)")
plt.title("Temperatura vs. Vibración (rojo = falla)")
plt.colorbar(label="Falla")
plt.show()
```

## Conclusiones del ejercicio

- El dataset es sintético y sirve como práctica del flujo: carga → resumen → agrupación → visualización.
- Se observa si la vibración media difiere entre registros con y sin falla.
